### Connections

```python
from dotenv import load_dotenv
from pathlib import Path
from pymongo import MongoClient
from pymongo.server_api import ServerApi
import certifi
import os
```

```python
# Load environment variables from .env file
env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)

MONGODB_URI = os.getenv("MONGODB_URI")
print("URI loaded?", MONGODB_URI is not None)

client = MongoClient(
    MONGODB_URI,
    server_api=ServerApi("1"),
    tls=True,
    tlsCAFile=certifi.where(),
    serverSelectionTimeoutMS=5000,
    connectTimeoutMS=5000,
    socketTimeoutMS=5000
)

try:
    client.admin.command("ping")
    print("✅ Successfully connected to MongoDB!")
except Exception as e:
    print("❌ Connection failed:")
    print(e)
```

```python
db = client.sample_mflix
movies = db.movies
print(f"📊 Database: {db.name}")
print(f"📁 Collection: movies ({movies.count_documents({})} documents)")
```


### Question 1

Question: From the `movies` collection, return the documents with the `plot` that starts with `"war"` in acending order of released date, print only title, plot and released fields. Limit the result to 5.


**Answer:**

```python
# ============================================================================
# Question 1
# ============================================================================

# ^war = plot starts with "war"
# "$options": "i" = ignore uppercase/lowercase
# projection = only show title, plot, released
# .sort("released", 1) = earliest released first
# .limit(5) = only 5 resul

query ={
    "plot": {"$regex": r"^war", "$options":"i"}
}

projection = {
    "_id": 0,
    "title": 1,
    "plot": 1,
    "released": 1
}

results = movies.find(query, projection).sort("released", 1).limit(5)

for doc in results:
    print(doc)
```

### Question 2

Question: Group by `rated` and count the number of movies in each.

**Answer:**

```python
# ============================================================================
# Question 2
# ============================================================================

pipeline = [
    {
        "$group": {
            "_id": "$rated",
            "count": {"$sum": 1}
        }
    },
    {
        "$sort": {
            "count": -1
        }
    }
]

results = movies.aggregate(pipeline)

for doc in results:
    print(doc)
```


### Question 3

Question: Count the number of movies with 3 comments or more.


**Answer:**


```python
# ============================================================================
# Question 3
# ============================================================================

# "num_mflix_comments" = the field that stores how many comments a movie has
# "$gte": 3 = greater than or equal to 3
# count_documents(query) = count how many movie documents match the condition
# print(count) = show the final numb

query = {
    "num_mflix_comments": {"$gte": 3}
}

count = movies.count_documents(query)

print(count)
```
